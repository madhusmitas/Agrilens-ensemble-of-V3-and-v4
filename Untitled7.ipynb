```python
!pip install -q datasets timm

from datasets import load_dataset
import torch, timm
from torchvision import transforms
from torch.utils.data import DataLoader
import torch.nn as nn
from google.colab import files

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)

ds = load_dataset("ButterChicken98/tomato_plantvillage_rag_k2_ablation", streaming=True)

tfm = transforms.Compose([
    transforms.Resize((192,192)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],[0.229,0.224,0.225])
])

def collate(b):
    x = torch.stack([tfm(i['image'].convert("RGB")) for i in b])
    y = torch.tensor([i['label'] for i in b])
    return x,y

loader = DataLoader(ds['train'], batch_size=64, collate_fn=collate)

model_v3 = timm.create_model("mobilenetv3_small_100", pretrained=True, num_classes=10).to(device)
model_v4 = timm.create_model("mobilenetv4_hybrid_medium", pretrained=True, num_classes=10).to(device)

for m in [model_v3, model_v4]:
    for p in m.parameters(): p.requires_grad = False
    for p in m.get_classifier().parameters(): p.requires_grad = True

def fast_train(model, name):
    opt = torch.optim.Adam(filter(lambda p: p.requires_grad, model.parameters()), lr=1e-3)
    crit = nn.CrossEntropyLoss()
    model.train()
    for step, (imgs, labels) in enumerate(loader):
        if step >= 80: break
        imgs, labels = imgs.to(device), labels.to(device)
        opt.zero_grad()
        loss = crit(model(imgs), labels)
        loss.backward()
        opt.step()
        if step%10==0: print(f"{name} Step {step} Loss {loss.item():.3f}")
    return model

print("Training V3...")
fast_train(model_v3, "V3")
print("Training V4...")
fast_train(model_v4, "V4")

# SAVE
torch.save(model_v3.to("cpu").half().state_dict(), "/content/v3_3MB.pt", _use_new_zipfile_serialization=True)
torch.save(model_v4.to("cpu").half().state_dict(), "/content/v4_4MB.pt", _use_new_zipfile_serialization=True)
print("SAVED! Now auto-downloading...")

# AUTO DOWNLOAD - will pop up in mom's phone
files.download('/content/v3_3MB.pt')
files.download('/content/v4_4MB.pt')

print("DONE! Check Downloads folder in mom's phone")
```
